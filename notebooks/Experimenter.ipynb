```python
import pandas as pd

# Example data with many duplicate values
data = [1, 1, 1, 1, 2, 3, 4, 5, 5, 5, 5, 5, 5, 6, 7, 8, 9, 10, 10, 10, 10]
series = pd.Series(data)

# --- Scenario 1: Default behavior (duplicates="raise") ---
try:
    # This might raise an error if duplicate edges occur,
    # as it's the default behavior (or you could explicitly use duplicates="raise")
    bins_default = pd.qcut(series, q=4)  # Try to create 4 quartiles
    print(
        "Bins with default behavior (if it didn't error):\n",
        bins_default.value_counts(),
    )
except ValueError as e:
    print(f"Error without 'duplicates=\"drop\"': {e}")
    print(
        "Notice the error about 'Bin edges must be unique'. This is because trying to split this data into 4 equal quantiles results in non-unique boundaries due to the many identical values."
    )
    print("\n")

# --- Scenario 2: Using duplicates="drop" ---
bins_dropped = pd.qcut(series, q=4, duplicates="drop")
print("Bins with 'duplicates=\"drop\"':\n", bins_dropped.value_counts().sort_index())
print(f"Number of unique bins created: {len(bins_dropped.cat.categories)}")
print(
    "Notice that even though we asked for q=4 (quartiles), we got fewer unique bins because the problematic duplicate edges were dropped."
)
```
